```python
from dotenv import load_dotenv

load_dotenv()

import pandas as pd
```

```python
documents = [
"Machine Learning is a subset of artificial intelligence",
"Deep Learning uses neural networks with multiple layers",
"Python is a programming language popular for data science",
"India won the cricket world cup in 2011",
"AI models can understand and generate human language"
]
```

```python
from langchain_mistralai import MistralAIEmbeddings
```

```python
embeddings = MistralAIEmbeddings(model="mistral-embed")
```

```python
vector = embeddings.embed_documents(documents)
len(vector)
```

```python
# len(vector[0]) # 1024
# print(vector)
df = pd.DataFrame(vector) #for this first we have to import pandas as pd
display(df)
```

# Store Vector Embeddings in Vector Databases

##### use ChromaDB to store vector embeddings and retrieve them using semantic search


```python
from langchain_chroma import Chroma
```

```python
vectorStore = Chroma.from_texts(
    texts=documents,
    embedding=embeddings,
    persist_directory="./vector_db"
)
```

```python
query = "covid-19"
result = vectorStore.similarity_search(query=query , k=2) # k means how many similar documents I want to retrieve
result
```

```python
query = "covid-19"
result = vectorStore.similarity_search_with_score(query=query, k=2)  # k means how many similar documents I want to retrieve

for doc, score in result:
    print(f"Score: {score} | Content: {doc.page_content}")
```
