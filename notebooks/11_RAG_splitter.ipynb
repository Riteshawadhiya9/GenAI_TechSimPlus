# Document Text Loading and Splitting

```python
from langchain_community.document_loaders import TextLoader , PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
loader = TextLoader('../data/RAG_notes.txt')
text = loader.load()
```

```python
pdfLoader = PyPDFLoader("../data/PharmNest_MVP.pdf")
pdfText = pdfLoader.load()
len(pdfText)
```

```python
splitters = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=450)
```

```python
text_chunks = splitters.split_documents(text)
len(text_chunks)
```

```python
text_chunks
```

```python
pdf_chunks = splitters.split_documents(pdfText)
len(pdf_chunks)
```
